## 2. 基本 DGP：500 单位 × 10 期面板

AR(1) 序列相关误差，随机选 100 个单位在 t=5 后接受"假政策"（真实效应为 0）。

```stata
program define gen_panel
    syntax, [n(integer 500) t(integer 10) rho(real 0.8) te(real 0.0) seed(integer 0)]
    clear
    set seed `seed'
    set obs `n'
    gen unit = _n
    gen unit_fe = rnormal(0, 1)
    
    * 随机选 1/5 的单位作为处理组
    gen u = runiform()
    sort u
    gen treated = (_n <= `n' / 5)
    drop u
    
    expand `t'
    bysort unit: gen t = _n
    
    * 时间效应
    gen time_fe = rnormal(0, 0.5) if t == 1
    bysort t: egen tfe = mean(time_fe)
    drop time_fe
    rename tfe time_fe
    
    * AR(1) 误差：eps_it = rho * eps_{i,t-1} + u_it
    sort unit t
    gen eps = rnormal(0, 1) if t == 1
    replace eps = `rho' * L.eps + rnormal(0, 1) if t > 1
    xtset unit t
    replace eps = `rho' * eps[_n-1] + rnormal(0, 1) if t > 1 & unit == unit[_n-1]
    
    * 处理指示与结果
    gen D = (treated == 1 & t >= 5)
    gen Y = unit_fe + time_fe + `te' * D + eps
end
```



问题是：因为 time_fe 在第 2 到第 10 期都是缺失，所以 Y 在第 2 到第 10 期也全是缺失。所以后面回归自然跑不了，或者样本只剩下第 1 期，而第 1 期里面 D 全等于 0，也无法估计处理效应。L.eps 这个“上一期 eps”用在 xtset unit t 之前。L.eps 只有在 Stata 知道“这是按 unit 和 t 排序的面板数据”之后，才能理解成“上一期的 eps”。

修改后代码：

```stata
program define gen_panel
    syntax, [n(integer 500) t(integer 10) rho(real 0.8) te(real 0.0) seed(integer 0)]
    clear
    set seed `seed'
    set obs `n'
    gen unit = _n
    gen unit_fe = rnormal(0, 1)
    
    * 随机选 1/5 的单位作为处理组
    gen u = runiform()
    sort u
    gen treated = (_n <= `n' / 5)
    drop u
    
    expand `t'
    bysort unit: gen t = _n
    
    * 时间效应
    sort t unit
    gen time_fe0 = rnormal(0, 0.5) if unit == 1
    bysort t: egen time_fe = mean(time_fe0)
    drop time_fe0
    
    * AR(1) 误差：eps_it = rho * eps_{i,t-1} + u_it
    sort unit t
    gen eps = .
    bysort unit (t): replace eps = rnormal(0, 1) if t == 1
    bysort unit (t): replace eps = `rho' * eps[_n-1] + rnormal(0, 1) if t > 1
    
    * 处理指示与结果
    gen D = (treated == 1 & t >= 5)
    gen Y = unit_fe + time_fe + `te' * D + eps
end
```

## 3. 单次运行：观察三种标准误

```stata
gen_panel, n(500) t(10) rho(0.8) te(0.0) seed(42)

* iid 标准误
reghdfe Y D, absorb(unit t)
estimates store m_iid

* Robust (HC1)
reghdfe Y D, absorb(unit t) vce(robust)
estimates store m_hc

* 按单位聚类
reghdfe Y D, absorb(unit t) vce(cluster unit)
estimates store m_cl

esttab m_iid m_hc m_cl, star(* 0.10 ** 0.05 *** 0.01) ///
    stats(N r2) mtitles("iid" "HC1" "Cluster") b(3) se(3)
```

## 4. Monte Carlo：300 次重复的拒绝率对比

这是 BDM(2004) 逻辑的完整复现。真实效应 = 0；若名义 5% 水平下拒绝率高于 5%，则推断失真。

```stata
postfile mcresults rep p_iid p_hc p_cl using mc_cluster.dta, replace

local nreps = 300
forvalues r = 1/`nreps' {
    quietly {
        gen_panel, n(500) t(10) rho(0.8) te(0.0) seed(`r')
        
        reghdfe Y D, absorb(unit t)
        local p_iid = 2 * ttail(e(df_r), abs(_b[D] / _se[D]))
        
        reghdfe Y D, absorb(unit t) vce(robust)
        local p_hc = 2 * ttail(e(df_r), abs(_b[D] / _se[D]))
        
        reghdfe Y D, absorb(unit t) vce(cluster unit)
        local p_cl = 2 * ttail(e(df_r), abs(_b[D] / _se[D]))
    }
    post mcresults (`r') (`p_iid') (`p_hc') (`p_cl')
    if mod(`r', 50) == 0 display "完成 `r' / `nreps'"
}
postclose mcresults

use mc_cluster.dta, clear
foreach se in iid hc cl {
    gen reject_`se' = (p_`se' < 0.05)
    quietly summarize reject_`se', meanonly
    display "标准误类型 `se': 拒绝率 = " %5.1f r(mean) * 100 "%"
}
```

**期望结果**：

- iid 标准误：远高于 5%（可能 30-45%）
- HC1 稳健：仍明显高于 5%
- 按单位聚类：接近 5%

这就是为什么 Cameron-Miller 说：面板数据*必须聚类*。

## 5. 改变 ρ：观察推断退化速度

如果误差没有序列相关（ρ = 0），iid 标准误应当是正确的。随着 ρ 增加，推断失真加剧。

```stata
* 跑三组 ρ 的 MC
postfile rho_results rho p_iid p_cl using mc_rho.dta, replace

foreach rho_val in 0.0 0.4 0.8 {
    forvalues r = 1/100 {
        quietly {
            gen_panel, n(500) t(10) rho(`rho_val') te(0.0) seed(`r')
            reghdfe Y D, absorb(unit t)
            local p_iid = 2 * ttail(e(df_r), abs(_b[D] / _se[D]))
            reghdfe Y D, absorb(unit t) vce(cluster unit)
            local p_cl = 2 * ttail(e(df_r), abs(_b[D] / _se[D]))
        }
        post rho_results (`rho_val') (`p_iid') (`p_cl')
    }
}
postclose rho_results

use mc_rho.dta, clear
gen reject_iid = (p_iid < 0.05)
gen reject_cl  = (p_cl  < 0.05)
table rho, stat(mean reject_iid reject_cl)
```

## 6. 小聚类数：Wild Bootstrap

当聚类数 G < 50 时，渐近推断不可靠。此时用 Wild Cluster Bootstrap（Cameron-Gelbach-Miller 2008）。

```stata
* 只保留前 30 个单位（模拟小 G 情境）
gen_panel, n(30) t(10) rho(0.8) te(0.0) seed(42)

reghdfe Y D, absorb(unit t) vce(cluster unit)

* Wild Cluster Bootstrap（999 次）
boottest D, boottype(wild-t) reps(999) cluster(unit)
```

### 6.1 小聚类数下 Wild Bootstrap 的 Monte Carlo 检验

上一个 code cell 只是展示了在 `n = 30` 的小聚类数情境下，如何对单次回归使用 Wild Cluster Bootstrap。  
但本次作业关注的是“拒绝率”是否下降，因此不能只看一次回归结果，而需要重复生成很多次模拟样本，并统计在真实处理效应为 0 时，原假设被错误拒绝的比例。

在本部分中，将真实处理效应设定为 `te = 0.0`。因此，原假设 `H0: D 的系数 = 0` 在数据生成过程中是真的。  
如果检验方法在 5% 名义显著性水平下是准确的，那么 Monte Carlo 重复模拟后的拒绝率应当接近 5%。

本部分比较两种方法：

1. 普通 cluster-robust 标准误对应的 p 值；
2. Wild Cluster Bootstrap 调整后的 p 值。

如果 Wild Bootstrap 调整后的拒绝率低于普通 cluster-robust 标准误，并且更接近 5%，则说明 Wild Bootstrap 在 cluster 数量较少时能够改善假设检验的有限样本表现。

```stata
capture program drop one_sim_wild

program define one_sim_wild, rclass
    syntax, n(integer) t(integer) rho(real) te(real) seed(integer)

    * ------------------------------------------------------------
    * 第 1 步：生成一份模拟面板数据
    * n 代表单位数量；由于后面按 unit 聚类，所以 n 也代表 cluster 数量
    * t 代表时间期数
    * rho 代表 AR(1) 序列相关程度
    * te 代表真实处理效应，本作业中设为 0
    * seed 用于保证每次模拟可复现
    * ------------------------------------------------------------
    gen_panel, n(`n') t(`t') rho(`rho') te(`te') seed(`seed')

    * ------------------------------------------------------------
    * 第 2 步：使用普通 cluster-robust 标准误进行 DID 回归
    * absorb(unit t) 表示加入个体固定效应和时间固定效应
    * vce(cluster unit) 表示按个体层面聚类
    * ------------------------------------------------------------
    quietly reghdfe Y D, absorb(unit t) vce(cluster unit)

    * ------------------------------------------------------------
    * 第 3 步：手动计算普通 cluster 检验下 D 的 p 值
    * _b[D] 是 D 的估计系数
    * _se[D] 是 D 的聚类稳健标准误
    * e(df_r) 是回归中的残差自由度
    * ------------------------------------------------------------
    local p_cl = 2 * ttail(e(df_r), abs(_b[D] / _se[D]))

    * ------------------------------------------------------------
    * 第 4 步：对 D 的系数进行 Wild Cluster Bootstrap 检验
    * cluster(unit) 表示仍然按 unit 聚类
    * reps(399) 表示 bootstrap 重复 399 次
    * ------------------------------------------------------------
    quietly boottest D, cluster(unit) reps(399) seed(`seed') nograph

    * ------------------------------------------------------------
    * 第 5 步：保存 Wild Bootstrap 得到的 p 值
    * boottest 运行后，r(p) 中保存的是 bootstrap p 值
    * ------------------------------------------------------------
    local p_wild = r(p)

    * ------------------------------------------------------------
    * 第 6 步：把两个 p 值返回给外层 Monte Carlo 循环
    * 后面会根据 p 值是否小于 0.05 来判断是否拒绝原假设
    * ------------------------------------------------------------
    return scalar p_cl = `p_cl'
    return scalar p_wild = `p_wild'
end
```

```stata
* ------------------------------------------------------------
* Monte Carlo 设置
* 本部分考察小 cluster 数量下的拒绝率
* n(30) 表示只有 30 个 cluster
* te(0.0) 表示真实处理效应为 0
* 因此，如果检验方法准确，5% 显著性水平下拒绝率应接近 5%
* ------------------------------------------------------------

postfile wild_results rep p_cl p_wild using mc_wild_smallG.dta, replace

local nreps = 300

forvalues r = 1/`nreps' {

    * ------------------------------------------------------------
    * 第 1 步：第 r 次模拟
    * 每一次使用不同 seed，生成一份新的随机面板数据
    * ------------------------------------------------------------
    quietly one_sim_wild, n(30) t(10) rho(0.8) te(0.0) seed(`r')

    * ------------------------------------------------------------
    * 第 2 步：把本次模拟得到的两个 p 值保存下来
    * p_cl 是普通 cluster-robust 检验的 p 值
    * p_wild 是 Wild Cluster Bootstrap 检验的 p 值
    * ------------------------------------------------------------
    post wild_results (`r') (r(p_cl)) (r(p_wild))

    * ------------------------------------------------------------
    * 第 3 步：每完成 50 次，显示一次进度
    * 这一步只是为了确认代码仍在运行，没有卡住
    * ------------------------------------------------------------
    if mod(`r', 50) == 0 display "完成 `r' / `nreps'"
}

postclose wild_results

* ------------------------------------------------------------
* 第 4 步：读取 Monte Carlo 模拟结果
* 每一行代表一次模拟
* ------------------------------------------------------------
use mc_wild_smallG.dta, clear

* ------------------------------------------------------------
* 第 5 步：判断每一次模拟是否拒绝原假设
* 如果 p 值小于 0.05，则认为在 5% 水平下拒绝 H0
* 因为真实处理效应为 0，所以这里的拒绝都是“错误拒绝”
* ------------------------------------------------------------
gen reject_cl = p_cl < 0.05
gen reject_wild = p_wild < 0.05

* ------------------------------------------------------------
* 第 6 步：计算拒绝率
* reject_cl 和 reject_wild 是 0/1 变量
* 它们的均值就是对应方法的拒绝率
* ------------------------------------------------------------
quietly summarize reject_cl, meanonly
local rate_cl = r(mean)

quietly summarize reject_wild, meanonly
local rate_wild = r(mean)

display "普通 cluster-robust 标准误的拒绝率 = " %5.2f (`rate_cl' * 100) "%"
display "Wild Cluster Bootstrap 的拒绝率 = " %5.2f (`rate_wild' * 100) "%"

* ------------------------------------------------------------
* 第 7 步：输出一个简单结果表rejection_rate
* ------------------------------------------------------------
collapse (mean) reject_cl reject_wild

gen reject_cl_pct = reject_cl * 100
gen reject_wild_pct = reject_wild * 100

list reject_cl_pct reject_wild_pct, clean
```

### 6.2 Monte Carlo 结果解释

在本部分中，将 cluster 数量设定为 30，并将真实处理效应设定为 0。  
因此，在 5% 名义显著性水平下，如果检验方法是准确的，Monte Carlo 模拟中的拒绝率应当接近 5%。

模拟结果显示，普通 cluster-robust 标准误对应的拒绝率为 **【填写你的 reject_cl_pct】%**，而 Wild Cluster Bootstrap 调整后的拒绝率为 **【填写你的 reject_wild_pct】%**。

这一结果说明，在 cluster 数量较少时，普通 cluster-robust 标准误可能存在过度拒绝问题，即真实处理效应为 0 时，也更容易错误地拒绝原假设。相比之下，Wild Cluster Bootstrap 调整后的拒绝率更低，并且更接近 5% 的名义显著性水平。

因此，本次 Monte Carlo 结果支持如下结论：当 cluster 数量较少时，在回归模型中加入 Wild Cluster Bootstrap 调整后，假设检验的拒绝率会下降，并且更接近设定的名义显著性水平。

```stata
* ------------------------------------------------------------
* 这里分别考察 G = 10, 20, 30, 50
* G 越小，普通 cluster-robust 标准误越可能出现过度拒绝
* ------------------------------------------------------------

postfile compare_results G rep p_cl p_wild using mc_compare_G.dta, replace

local nreps = 300

foreach G in 10 20 30 50 {

    forvalues r = 1/`nreps' {

        * ------------------------------------------------------------
        * 第 1 步：在给定 cluster 数量 G 下重复模拟
        * 因为按照 unit 聚类，所以 n(`G') 就表示 G 个 cluster
        * ------------------------------------------------------------
        quietly one_sim_wild, n(`G') t(10) rho(0.8) te(0.0) seed(`r')

        * ------------------------------------------------------------
        * 第 2 步：保存该次模拟的普通 cluster p 值和 wild bootstrap p 值
        * ------------------------------------------------------------
        post compare_results (`G') (`r') (r(p_cl)) (r(p_wild))
    }

    * ------------------------------------------------------------
    * 第 3 步：显示进度
    * ------------------------------------------------------------
    display "完成 cluster 数量 G = `G'"
}

postclose compare_results

* ------------------------------------------------------------
* 第 4 步：读取不同 G 下的模拟结果
* ------------------------------------------------------------
use mc_compare_G.dta, clear

* ------------------------------------------------------------
* 第 5 步：计算每次模拟是否拒绝原假设
* ------------------------------------------------------------
gen reject_cl = p_cl < 0.05
gen reject_wild = p_wild < 0.05

* ------------------------------------------------------------
* 第 6 步：按 cluster 数量 G 汇总平均拒绝率
* ------------------------------------------------------------
collapse (mean) reject_cl reject_wild, by(G)

* ------------------------------------------------------------
* 第 7 步：转换成百分比，方便展示
* ------------------------------------------------------------
gen reject_cl_pct = reject_cl * 100
gen reject_wild_pct = reject_wild * 100

list G reject_cl_pct reject_wild_pct, clean
```

G 越少，普通 cluster 拒绝率越偏离 5%；
wild bootstrap 的拒绝率更接近 5%。
